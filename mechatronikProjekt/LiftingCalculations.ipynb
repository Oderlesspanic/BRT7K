```python
import numpy as np

# motor parameters
C_Torque_Motor  =   9.17 * 10**(-3)         # Nm/A
C_Speed_Motor   =   1040                    # rpm/V
n_Motor         =   0.52                    # 1     -> Efficiency

nominal_voltage =   12                      # V
nominal_current =   0.293                   # A

speed_Motor     =   C_Speed_Motor * nominal_voltage     # rpm
torque_Motor = C_Torque_Motor * nominal_current * n_Motor # Nm

speed_Motor = 5460                          # rpm, given by the manufacturer
torque_Motor = 2.34 * 10**(-3)              # Nm, given by the manufacturer

# gear parameters
i_planetarygear =   157                     # 1     -> gearratio
n_planetarygear =   0.73                    # 1     -> Efficiency

# Complete Drive Parameters

speed_Drive     =   speed_Motor/i_planetarygear     # rpm
torque_Drive    =   torque_Motor * i_planetarygear * n_planetarygear  # Nm
power_Drive     =   (2*np.pi*torque_Drive*speed_Drive)/(60)     # W

print(f"max output Torque Drive: {torque_Drive*1000:.2f}mNm, with speed: {speed_Drive:.2f}rpm, overall Mechanical output Power: {power_Drive:.2f}W")

# Calculate Needed Torque to Lift the Gripper and Object!
n_mechanical_Construction   =   0.5         # 1     -> Efficiency of the Belt Lifting mechanism (guessed!)
lifting_speed_mm            =   5                      # mm/s


belt_pitch      =    5                                   # mm
min_possible_teeth_lifter    =   10                      # the min possible teeth count for the Lifting pulley on a 4mm shaft
diameter_lifter_pulley     =   (min_possible_teeth_lifter * belt_pitch) / np.pi    # mm                          # mm

lift_per_revolution =   belt_pitch * min_possible_teeth_lifter     # mm/rev
lifting_speed_rpm =   (lifting_speed_mm / lift_per_revolution) * 60

print(f"needed lifting speed in rpm: {lifting_speed_rpm:.2f}rpm for {lifting_speed_mm}mm/s with {min_possible_teeth_lifter} teeth on the lifting pulley")

# Calculate needed i of the drive to the shaft of the lifting pulley
i_needed = speed_Drive / lifting_speed_rpm
print(f"needed gear ratio to achieve the lifting speed: {i_needed:.2f}")

# Calculate the given torque on the lifting pulley
torque_lifting_pulley = torque_Drive * i_needed * n_mechanical_Construction
print(f"Torque on the lifting pulley: {torque_lifting_pulley*1000:.2f}mNm with the needed gear ratio of {i_needed:.2f} and the mechanical construction efficiency of {n_mechanical_Construction:.2f}")
print(f"Max possible weight to lift: {torque_lifting_pulley/((diameter_lifter_pulley/2)/1000)/9.81:.2f}kg")
```
